```python
import ollama
from ollama import chat
from ollama import ChatResponse
from langchain_core.prompts import ChatPromptTemplate
from langchain_ollama.llms import OllamaLLM
from pydantic import BaseModel, Field
from langchain_core.output_parsers import PydanticOutputParser
from langchain_ollama import ChatOllama
```

```python
response: ChatResponse = chat(model='smollm', messages=[
  {
    'role': 'user',
    'content': 'Why is the sky blue?',
  },
])
print(response['message']['content'])
# or access fields directly from the response object
print(response.message.content)
```

```python
template = """ Classify the sentiment of the following tweet as Positive, Negative, or Neutral:\n\n
Tweet: {tweet}
\n\n
Sentiment:
"""

prompt = ChatPromptTemplate.from_template(template)

model = OllamaLLM(model="smollm")
```

```python
chain = prompt | model

tweet = "I love the new features in this product! Amazing work by the team."

chain.invoke({"tweet": tweet})
```

## extracting speciific values from LLMs

```python
# Define your desired data structure.
class Sentiment(BaseModel):
    sentiment: str = Field(description="tweet sentiment")
```

```python
llm = ChatOllama(model="smollm")
structured_llm = llm.with_structured_output(Sentiment)
```

```python
chain = prompt | structured_llm

tweet = "I love the new features in this product! Amazing work by the team."

chain.invoke({"tweet": tweet})
```
